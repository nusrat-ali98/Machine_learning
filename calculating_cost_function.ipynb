```python
import numpy as np
%matplotlib widget
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500,0])
```

```python
def compute(x,y,w,b):
    m = x.shape[0]
    cost_sum = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost = (f_wb - y[i])**2
        cost_sum = cost_sum + cost
    total_cost =  (1/(2 * m))* cost_sum
    return total_cost
        
```

```python
t_f_wb = compute(x_train,y_train,200,100)
print(t_f_wb)
```
