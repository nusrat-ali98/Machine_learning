```python
import numpy as np
import math, copy
import pandas as ps
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1,2])
y_train = np.array([300,500])
```

```python
def cost_func(x,y,w,b):
    m = x.shape[0]
    cost = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost = cost + (f_wb - y[i]) **2
    total_cost = (1 / (2 * m) ) * cost
    return total_cost
        
```

```python
def compute_gradient(x,y,w,b):
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = w * x[i] + b
        dj_dw_i = (f_wb - y[i]) * x[i]
        dj_db_i = (f_wb - y[i])
        dj_dw += dj_dw_i
        dj_db += dj_db_i

    dj_dw = dj_dw/m
    dj_db = dj_db/m

    return dj_dw, dj_db
        

        
```

```python
def gradient_decent(x,y , w_in, b_in, alpha, num_itera, cost_function, cost_gradient):

    j_hist = []
    p_hist = []
    b = b_in
    w = w_in

    for i in range(num_itera):
        dj_dw,dj_db= cost_gradient(x,y,w,b)

        b = b - alpha * dj_db
        w = w - alpha * dj_dw


        if i < 100000:
            j_hist.append(cost_func(x,y,w,b))
            p_hist.append([w,b])

        if i % math.ceil(num_itera/10)==0:
            print(f"Iteration {i:4} : Cost {j_hist[-1]:2e}",
                  f"dj_dw:{ dj_dw:0.3e} : dj_db:{ dj_db:0.3e}",
                  f"w: {w:0.3e} : b: {b : 0.5e}")
    return w,b, j_hist,p_hist

    
```

```python
w_init = 0
b_init = 0
iteration = 10000
tmp_alpha = 1.0e-2

w_final, b_final , j_hist, p_hist = gradient_decent(x_train,y_train,w_init,b_init, tmp_alpha,iteration, cost_func,compute_gradient)
print(f"(w,b) found by gradient decent: ({w_final:8.4f},{b_final:8.4f})")
```
